```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

# Resolve the dataset path from the current working directory or its parents
filename = "diabetes_binary_smotenc_enn.csv"
# filename = "diabetes_binary_health_indicators_BRFSS2015.csv"
search_roots = [Path.cwd(), *Path.cwd().parents]
raw_csv = None

for root in search_roots:
    candidate = root / "data" / "processed" / filename
    if candidate.exists():
        raw_csv = candidate
        break

if raw_csv is None:
    raise FileNotFoundError(
        f"Data file not found. Checked: {[str(r / 'data' / 'raw' / filename) for r in search_roots]}"
    )

# Load the dataset
try:
    df = pd.read_csv(raw_csv)
except Exception as exc:
    raise RuntimeError(f"Failed to load dataset from {raw_csv}: {exc}") from exc

# Detect target column
candidate_targets = ["target", "label", "class", "y"]
target_col = 'diabetes_binary'

# Plot target distribution as an attractive donut chart
# Plot target distribution with a modern white-and-blue donut chart
counts = df[target_col].value_counts(dropna=False).sort_index()
labels = counts.index.astype(str).tolist()
values = counts.values

colors = ["#BFDBFE", "#2563EB"]  # light blue, primary blue

fig, ax = plt.subplots(figsize=(8, 6), facecolor="white")
ax.set_facecolor("white")

wedges, _, autotexts = ax.pie(
    values,
    startangle=90,
    counterclock=False,
    colors=colors,
    autopct="%1.1f%%",
    pctdistance=0.76,
    wedgeprops={
        "width": 0.42,
        "edgecolor": "white",
        "linewidth": 4,
    },
    textprops={
        "fontsize": 20,          # percentage labels
        "fontweight": "bold",
        "color": "#1E3A5F",
    },
)

ax.text(
    0, 0.10, "Total records",
    ha="center", va="center",
    fontsize=17,
    color="#64748B",
)
ax.text(
    0, -0.12, f"{values.sum():,}",
    ha="center", va="center",
    fontsize=32,
    fontweight="bold",
    color="#0F172A",
)

ax.set_title(
    "Diabetes Class Distribution",
    fontsize=24,
    fontweight="bold",
    color="#0F172A",
    pad=30,
)

ax.legend(
    wedges,
    [f"Class {label}: {value:,}" for label, value in zip(labels, values)],
    title="Diabetes_binary",
    loc="lower center",
    bbox_to_anchor=(0.5, -0.14),
    ncol=2,
    frameon=False,
    fontsize=20,
    title_fontsize=16,
)
```
