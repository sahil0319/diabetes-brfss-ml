```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

# Resolve the dataset path from the current working directory or its parents
filename = "diabetes_binary_health_indicators_BRFSS2015.csv"
search_roots = [Path.cwd(), *Path.cwd().parents]
raw_csv = None

for root in search_roots:
    candidate = root / "data" / "raw" / filename
    if candidate.exists():
        raw_csv = candidate
        break

if raw_csv is None:
    raise FileNotFoundError(
        f"Data file not found. Checked: {[str(r / 'data' / 'raw' / filename) for r in search_roots]}"
    )

# Load the dataset
try:
    df = pd.read_csv(raw_csv)
except Exception as exc:
    raise RuntimeError(f"Failed to load dataset from {raw_csv}: {exc}") from exc

# Detect target column
candidate_targets = ["target", "label", "class", "y"]
target_col = 'Diabetes_binary'

# Plot target distribution as an attractive donut chart
counts = df[target_col].value_counts(dropna=False)
labels = [f"{label}" for label in counts.index.astype(str)]
values = counts.values

plt.style.use("seaborn-v0_8-darkgrid")
fig, ax = plt.subplots(figsize=(8, 8))

colors = ["#4C78A8", "#F58518"]
ax.pie(
    values,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90,
    pctdistance=0.7,
    wedgeprops={"width": 0.6, "edgecolor": "white", "linewidth": 1.2},
    colors=colors,
    textprops={"fontsize": 11, "color": "#1f2937"},
)

ax.set_title(
    f"Target Class Distribution ({target_col})",
    fontsize=15,
    fontweight="bold",
    pad=20,
)
ax.axis("equal")

plt.tight_layout()
plt.show()

print(f"Loaded: {raw_csv}")
print(f"Target column used: {target_col}")
```
